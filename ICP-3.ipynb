```python
import pandas as pd
```

```python
# Load the dataset (assuming you've uploaded the CSV)
df = pd.read_csv('/content/Sacramento-RealEstate-Transactions.csv')
```

```python
# 1. Number of columns in the DataFrame
num_columns = len(df.columns)
print("Number of columns:", num_columns)
```

```python
# 2. Mean of house prices
mean_price = df['price'].mean()
print("Mean house price:", mean_price)
```

```python
# 4. Different types of property in the dataset
property_types = df['type'].unique()
print("Different property types:", property_types)
```

```python
# 5. Maximum price of a "condo"
max_condo_price = df[df['type'].str.lower() == 'condo']['price'].max()
print("Maximum price of a condo:", max_condo_price)
```

```python
# 6. Mean price of each type of property
mean_price_by_type = df.groupby('type')['price'].mean()
print("Mean price by property type:\n", mean_price_by_type)
```

```python
# 7. Count of different property types
property_type_counts = df['type'].value_counts()
print("Property type counts:\n", property_type_counts)
```
